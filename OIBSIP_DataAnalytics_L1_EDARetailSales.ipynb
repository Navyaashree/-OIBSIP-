```python
df.duplicated().sum()
df.isnull().sum()
df.columns
df.groupby('Category')['Amount'].sum().sort_values(ascending=False).plot(kind='bar',color='green')
plt.title('Total Amount by Category')
plt.xlabel('Category')
plt.ylabel('Amount')
plt.show()

```

```python
sns.scatterplot(data=df,x='PreviousPurchases',y='DiscountApplied(%)',color='orange')
plt.title('Discount over previous purchases')
plt.show()
```

```python
df.columns
```

```python
df.groupby('PaymentMethod')['Amount'].sum().plot(kind='pie',autopct='%1.1f%%')
plt.title('Amount by payment method')
plt.show()
```

```python
df.groupby('Gender')['Amount'].sum().plot(kind='pie',autopct='%1.1f%%')
plt.title('Total amount by gender')
plt.show()
```

```python
itempurchased_by_amount=df.groupby('ItemPurchased')['Amount'].sum().sort_values(ascending=False).head(10)
sns.barplot(data=itempurchased_by_amount,color='green')
plt.title('Top 10 items by amount ')
plt.xlabel('Items')
plt.ylabel('Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
df.columns
```

```python
df['Age_Group']=df['Age'].apply(lambda age:'child' if  age<18
                                 else 'Adult' if age<35
                                else 'Senior')
```

```python
df.head()
```

```python
df.groupby('Age_Group')['ItemRating'].mean().plot(kind='bar')
plt.title('Avg rating by age group')
plt.xlabel('Age Group')
plt.ylabel('Ratings')
plt.xticks(rotation=0)
plt.show()
```

```python
sns.heatmap(df[['Amount','ItemPurchased', 'DiscountApplied(%)', 'PreviousPurchases']].corr(),annot=True,cmap='coolwarm')
plt.title('Correlation of Numeric Columns')
plt.show()
```
